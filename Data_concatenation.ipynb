```python
import pandas as pd 
```

```python
# Concatenation

df1 = pd.DataFrame({
    "id" : [1,2,3],
    "name" : ["Adam","Eve","Bob"]   
})

df2 = pd.DataFrame({
    "id" : [4,5,6],
    "name" : ["Charlie","Fester","Gilian"]
})

pd.concat ([df1,df2]) #row wise
pd.concat ([df1,df2] , ignore_index = True) # with this ignore indeex we can get new labels and inore the old one

pd.concat([df1,df2],axis = 1)
```
