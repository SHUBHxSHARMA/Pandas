```python
import pandas as pd
```

```python
df = pd.read_csv("raw_data.csv")
print(df)
```

```python
# Cleaning Data - missing value

df.isnull()

df.isnull().sum()

df.dropna() # row drop 
df.dropna(axis = 1) # column drop 

df.fillna(0)

age_mean = df["age"].mean()

cleaned_data = df.copy()
age_mean = cleaned_data["age"].mean()
cleaned_data["age"]= cleaned_data["age"].fillna(age_mean)

df.ffill()

df.bfill()
```

```python
#handling duplicate value 

df.duplicated()
df[["country","gender"]].duplicated()

df.drop_duplicates()

```
