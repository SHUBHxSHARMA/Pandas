```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "country" : ["USA","USA" , "India", "India"],
    "year" : [2020 , 2021, 2020, 2021],
    "sales" : [100,120,90,110],
    "profit" : [20,25,18,22],

}) 
df
```

```python
# melt function - wide to long format
melted_df = df.melt(
    id_vars = ["country" , "year"] , #these columns will not change and will be preserved
    value_vars = ["sales","profit"] , #these are the values that we are changing here we r combing it into one column
    var_name = "metrics" , 
    value_name = "value"    
)
```

```python
# Pivot function - Long to wide format

original = melted_df.pivot(
    index = ["country","year"],
    columns = "metrics",
    values = "value"
)
original
```
