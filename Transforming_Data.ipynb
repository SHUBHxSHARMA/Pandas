```python
import pandas as pd
```

```python
df = pd.read_csv("raw_data.csv")
df
```

```python
# copying the data into new variable so tht we can do all the changes in this copied data
df2 = df.copy()
#writing the logic imside the apply function , also using lambda to write the whole function in jst one line
df2["tax"] = df2["income"].apply(lambda x : "20%" if x >= 50_000 else "10%" )
df2
```

```python
# mapping 
# here in data we want Male -> M , Femaile -> F , Unknown -> U so to do tht firste we create a dictionary

gender_map = {"Male": "M", "Female":"F" , "Unknown":"U" }
df2["gender"] = df2["gender"].map(gender_map)
df2
```

```python
# assigning
df2 = df2.assign(new_income = df2["income"] * 1.1)
df2
```

```python
# Replacing (old,new)

df2["country"] = df2["country"].replace("USA","US")
df2
```

```python
# Rename function

df2.columns = ["Id","Name","Age","Country","Gender","Income","Tax","New_Income"]

#changing old to new
df2 = df2.rename(columns={"Income":"Salary"})

df2.rename(index={1:"First"})
```

```python
# sort 
df2.sort_values("Salary")
df2.sort_values(["Salary","Age"])

# sorting is in accending order by default

sorted_df = df2.sort_values("Salary",ascending = False)

#here as we see the iindex is jumbled to make it sort we use the below function
sorted_df.sort_index()
```
