```python
import pandas as pd 
```

```python
df = pd.read_csv("IRIS.csv")
df
```

```python
# First 10 rows
# Shape and data type
# Summary statistics(mean,std,min,max)
print(df.head(10))
print(type(df),df.shape)
print(df.describe())
```

```python
# Select only those rows where :
# •petal_length>4.5
# •species="Iris-virginica

df[(df["petal_length"]>4.5)&(df["species"] == "Iris-virginica")]
```

```python
# group by species and compute:
#  average sepal_lenght
#  maximum petal_width
#  Standard deviation of sepal_width

print(df.groupby("species")["sepal_length"].mean())
print(df.groupby("species")["petal_length"].max())
print(df.groupby("species")["sepal_width"].std())

```

```python
# create a new column : "petal-ratio" = petal_length /petal-width
# for each species , find the average petal-ratio.

df["petal_ratio"] = df["petal_length"] / df["petal_width"]
df
```
