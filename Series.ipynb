```python
import pandas as pd
```

```python
s = pd.Series([1,2,3,4,5])
print(s)
print(type(s))

# index
print(s[0])
print(s[1])
```

```python
s = pd.Series([23,24,25,26] , index = ["Shubh","Shubranil","Tanay","Kailash"] )
print(s)
print(s["Shubh"])

# we can print all the index values by
print(s.index)
```

```python
# vectorization done in series

s1 = pd.Series([23,24,25,26])
s2 = pd.Series([1,2,3,4,5])
print(s1+s2)
```

```python
# Immutable property in series
s1 = pd.Series([23,24,25,26])
s1[0] = 1
print(s1[0])
```

```python
# Drop function in series
s1 = pd.Series([23,24,25,26])
changedseries = s1.drop([0])
print(s1)
print(changedseries)
```
