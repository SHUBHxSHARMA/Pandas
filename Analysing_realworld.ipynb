```python
# Selecting Data 

#selecting values of  columns
df["city"]
df[["city","aqi"]]

#selecting values of Rows
df.loc[2]  # here 0 is the label
df.loc[0:2] # we can do slicing as well[start idx : end idx inclusive]

df.iloc[2]
df.iloc[0:2] # ending index exclusive

# cells - rows , col
df.loc[0,"aqi"]
df.loc[0,["city","aqi"]]
df.loc[0:2,["city","aqi"]]

# Select single scalar value 
df.at[0,"city"]
```

```python

# change without affecting original data
cities = df["city"].copy()
cities[0] = "Mumbai" #copy

cities  # copy
```

```python
 # FIlTERING  of data
df[df["aqi"] > 100]

#multiple conditions
df[ (df["aqi"]>100) & (df["temperature"]>30) ]
df[df["aqi"]>100][["city" , "aqi"]]
```

```python
# Query method
aqi_val = 100
df.query("aqi > 100 & temperature > 30")[["city","aqi"]]
df.query("aqi > @aqi_val & temperature > 30")[["city","aqi"]]
```
